```python
name = input("Enter your name: ")
age = int(input("Enter your age: "))
course = input("Enter your course: ")
percentage = float(input("Enter your percentage: "))

print("\n==============================")
print("      DIGITAL INTRODUCTION")
print("================================")
print(f"Name       : {name}  ")
print(f"Age        : {age}  ")
print(f"Course     : {course}")
print(f"Percentage : {percentage}%")
print("==============================")
```

```python
birth_year = int(input("Enter your birth year: "))

current_year = 2026
age = current_year - birth_year

print(f"Your approximate age is {age} years.")
```

```python
price = float(input("Enter the price of the item: "))
quantity = int(input("Enter the quantity: "))

total = price * quantity

print(f"Total Bill = ₹{total:.2f}")
```

```python
recharge = float(input("Enter recharge amount: "))
gst_percentage = float(input("Enter GST percentage: "))

gst = recharge * gst_percentage / 100
final_amount = recharge + gst

print(f"Recharge Amount : ₹{recharge:.2f}")
print(f"GST             : ₹{gst:.2f}")
print(f"Final Amount    : ₹{final_amount:.2f}")
```

```python
celsius = float(input("Enter temperature in Celsius: "))

fahrenheit = (celsius * 9/5) + 32

print(f"Temperature in Fahrenheit = {fahrenheit:.2f}°F")
```

```python
principal = float(input("Enter principal amount: "))
rate = float(input("Enter rate of interest (%): "))
time = float(input("Enter time (years): "))

simple_interest = (principal * rate * time) / 100
total_amount = principal + simple_interest

print(f"Simple Interest = ₹{simple_interest:.2f}")
print(f"Total Amount   = ₹{total_amount:.2f}")
```

```python
speed = float(input("Enter speed (km/h): "))
time = float(input("Enter travel time (hours): "))

distance = speed * time

print(f"Distance Travelled = {distance:.2f} km")
```

```python
height = float(input("Enter height in metres: "))
weight = float(input("Enter weight in kilograms: "))

bmi = weight / (height ** 2)

print(f"Your BMI is {bmi:.2f}")
```

```python
units = float(input("Enter units consumed: "))
cost_per_unit = float(input("Enter cost per unit: ₹"))

bill = units * cost_per_unit

print(f"Units Consumed : {units}")
print(f"Cost Per Unit  : ₹{cost_per_unit:.2f}")
print(f"Estimated Bill : ₹{bill:.2f}")
```

```python
product1 = input("Enter first product name: ")
price1 = float(input("Enter price of first product: ₹"))

product2 = input("Enter second product name: ")
price2 = float(input("Enter price of second product: ₹"))

product3 = input("Enter third product name: ")
price3 = float(input("Enter price of third product: ₹"))

total = price1 + price2 + price3

print("\n==============================")
print("       SHOPPING RECEIPT")
print("==============================")
print(f"{product1:<15} ₹{price1:.2f}")
print(f"{product2:<15} ₹{price2:.2f}")
print(f"{product3:<15} ₹{price3:.2f}")
print("------------------------------")
print(f"Total Amount     ₹{total:.2f}")
print("==============================")
```
